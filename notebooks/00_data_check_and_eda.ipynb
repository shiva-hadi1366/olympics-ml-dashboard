# 00 — Data Check & EDA

This notebook loads `data/raw/athlete_events.zip` and performs basic checks and quick EDA that is useful for the Streamlit dashboard and the two-stage ML pipeline.

Run from the `notebooks/` folder.


```python
import zipfile
import pandas as pd
import numpy as np
from pathlib import Path

REPO_ROOT = Path('..')
RAW_ZIP = REPO_ROOT / 'data' / 'raw' / 'athlete_events.zip'

RAW_ZIP
```

```python
with zipfile.ZipFile(RAW_ZIP, 'r') as z:
    print('zip files:', z.namelist())
    with z.open('athlete_events.csv') as f:
        df = pd.read_csv(f)

df.columns = [c.strip().lower().replace(' ', '_') for c in df.columns]
df.shape
```

```python
df.head()
```

```python
# Missingness
missing = (df.isna().mean() * 100).sort_values(ascending=False)
missing.head(20)
```

```python
# Targets preview
df['has_medal'] = df['medal'].notna().astype(int)
df['medal_class'] = df['medal'].fillna('None')

df[['medal', 'has_medal', 'medal_class']].head(10)
```

```python
df['has_medal'].value_counts(), df['has_medal'].value_counts(normalize=True)
```

```python
df['medal_class'].value_counts()
```

```python
# Some quick aggregates for dashboard ideas
if set(['year', 'season', 'sport', 'event']).issubset(df.columns):
    display(df.groupby(['season']).size().rename('rows'))
    display(df.groupby(['year']).size().head().rename('rows'))
    display(df.groupby(['sport']).size().sort_values(ascending=False).head(15).rename('rows'))
```
